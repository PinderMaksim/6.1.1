```python
import matplotlib.pyplot as plt

def gnome_sort(arr):
    # Контроль правильності введення даних [cite: 29]
    if not isinstance(arr, list):
        raise TypeError("Потрібно ввести список чисел.")

    i = 0
    while i < len(arr):
        if i == 0 or arr[i] >= arr[i - 1]:
            i += 1
        else:
            arr[i], arr[i - 1] = arr[i - 1], arr[i]
            i -= 1
    return arr

# Тестові приклади [cite: 28]
print("Тест 1 (звичайний):", gnome_sort([34, 2, 10, -9]))
print("Тест 2 (відсортований):", gnome_sort([1, 2, 3, 4]))
print("Тест 3 (зворотний):", gnome_sort([5, 4, 3, 2, 1]))
```

```python
data = [12, 4, 5, 9, 1]
labels = ["A", "B", "C", "D", "E"]
sorted_data = gnome_sort(data.copy())

plt.figure(figsize=(8,4))
plt.bar(labels, sorted_data, color='green')
plt.title("Візуалізація відсортованих даних (Gnome Sort)")
plt.ylabel("Значення")
plt.show()
```
